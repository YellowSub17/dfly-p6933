```python
import numpy as np
import h5py
import matplotlib.pyplot as plt
```

```python
hit_cache_path = '../buchin/spb_dev/run_small_data/run0266_hit_cache.h5'

diag_cache_path = '../buchin/spb_dev/run_small_data/run0266_diagnostics.h5'
```

```python
with h5py.File(hit_cache_path,'r') as f:
    imgs = f['/images'][...]

with h5py.File(diag_cache_path,'r') as f:
    saxs_hs = f['/hitscore_saxs'][...]
    brag_hs = f['/hitscore_bragg'][...]

```
