```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

DRAGONFLY_RECON = './test_0003'


log_arr = np.loadtxt(f'{DRAGONFLY_RECON}/logs/EMC.log',dtype=float, skiprows=19)


i_iter, time, rms_change, info_rate, log_likelihood, num_rot, beta = log_arr.T
```

```python
fig, axes = plt.subplots(1, 2, sharex=True, figsize=(10,6))
axes[0].plot(i_iter, rms_change)
axes[0].set_yscale('log')
axes[0].set_ylabel('RMS change')
axes[0].set_xlabel('EMC Iteration')

axes[1].plot(i_iter, beta)
axes[1].set_ylabel('beta')
axes[0].set_xlabel('EMC Iteration')
```
